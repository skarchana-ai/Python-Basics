Sum of digits

```python
number = input("Type your number:")   #input() stores 1234 as text:"1234"
total=0   #starts the sum at zero

for digit in number:   #takes each character:1,2,3,4
  total += int(digit)   # int(digit) converts each charatecer into number
print(total)
```

*   int(input()) - when you need a whole number for calculations
*   input() - when you need to process the input charater by character.



Count the number of vowels in a given word

```python
word=input("Enter a word:")
count=0

for letter in word:
  if letter.lower() in "aeiou":    #letter.lower() handles both uppercase and lowercase letters.
    count +=1
print(count)
```

Find the largest number in a list

```python
numbers=[2,3,4,5,6]
print(max(numbers))    #built in max() function
```

```python
numbers =[2,3,4,5,6]
numbers.sort()       #sort() aranges num in ascending order
print(numbers[-1])    #numbers[-1] selects the last element, which is the largest
```

```python
#Without using built in functions
numbers =[2,3,6,5,4]

largest =0

for n in numbers:
  if n > largest:    #2>0, so largest=2; 3>2,so largest =3 ; 6>3, so largest =6; 5>6 is false,so largest remains 6 ; 4>6 is false, so largest remains 6
    largest = n
print(largest)
```

Find the second largest number

```python
numbers = [12, 35, 23, 45, 67, 34]

largest = float('-inf')
second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif largest > num > second_largest:
      second_largest = number
print(second_largest)
```

Write a program to count how many times each word appears in a sentence, ignoring uppercase and lowercase differences.

```python

sentence = "Python is easy and Python is useful"

words = sentence.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

Write a function that accepts a list of marks, calculates the average, and returns all marks greater than the average.

```python

def analyze_marks(marks):
    average = sum(marks) / len(marks)

    above_average = []

    for mark in marks:
        if mark > average:
            above_average.append(mark)

    return average, above_average


marks = [65, 80, 45, 90, 70]

average, result = analyze_marks(marks)

print("Average:", average)
print("Marks above average:", result)
```
